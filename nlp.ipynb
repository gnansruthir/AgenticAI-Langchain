```python
! python -m spacy download en_core_web_md

import spacy
en = spacy.load("en_core_web_md")
```

```python
text=("donald trump is the president of usa and 45 years old ruling democratic country")
print(text)
```

```python
doc_en=en(text)
```

# to display in tabulor format

```python
from IPython.display import HTML, display
import tabulate

tokens = [[token] for token in doc_en]
display(HTML(tabulate.tabulate(tokens, tablefmt="html")))
```

# parts of speech -done with each token

```python
features = [[t.orth_, t.lemma_, t.pos_, t.tag_] for t in doc_en]
display(HTML(tabulate.tabulate(features, tablefmt="html")))
```
